## Nhóm 1: Nhóm Mô hình Tuyến tính và Thống kê
Đại diện cho các phương pháp thống kê cổ điểm.
Các mô hình này cực kỳ đơn giản, trực quan, tốn rất ít tài nguyên tính toán và có tính giải thích toán học cao

```python
from fraud_models import get_tabnet_model, get_traditional_models

RANDOM_STATE = 42
traditional_models = get_traditional_models(random_state=RANDOM_STATE)
assert len(traditional_models) == 12, "Expected exactly 12 traditional models."
print(f"[READY] Loaded {len(traditional_models)} traditional models.")
```

```python
# Nhóm 1: Mô hình Tuyến tính và Thống kê
logistic_regression_model = traditional_models["Logistic_Regression"]
lda_model = traditional_models["LDA"]

group_1_models = {
    "Logistic_Regression": logistic_regression_model,
    "LDA": lda_model,
}

for model_name, model in group_1_models.items():
    model_path = f"{type(model).__module__}.{type(model).__name__}"
    print(f"[READY] {model_name}: {model_path}")
```

## Nhóm 2: Nhóm Mô hình Cây quyết định đơn lẻ
Đại diện cho phương pháp phân loại dựa trên các tập luật logic "NẾU - THÌ" (IF-THEN) phân cấp. Mô hình này mô phỏng trực quan quy trình ra quyết định của con người

```python
# Nhóm 2: Mô hình Cây quyết định đơn lẻ
decision_tree_model = traditional_models["Decision_Tree"]

group_2_models = {"Decision_Tree": decision_tree_model}

for model_name, model in group_2_models.items():
    model_path = f"{type(model).__module__}.{type(model).__name__}"
    print(f"[READY] {model_name}: {model_path}")
```

## Nhóm 3: Nhóm Mô hình Học kết hợp / Học tổ hợp
Là thuật toán mạnh mẽ nhất đối với dữ liệu dạng bảng trước khi học sâu phát triển mạnh.
Gồm hai trường phái 
### Trường phái Bagging:
xây dựng hàng trăm cây quyết định hoạt động hoàn toàn song song và độc lập trên các mẫu dữ liệu khác nhau, sau đó lấy trung bình kết quả dự đoán (hoặc bỏ phiếu) để đưa ra quyết định cuối cùng
### Trường phái Boosting
Thay vì hoạt động độc lập, các cây quyết định được xây dựng nối tiếp nhau, trong đó cây sau tập trung học tập và sửa chữa trực tiếp các lỗi sai của cây phía trước

```python
# Nhóm 3A: Bagging
random_forest_model = traditional_models["Random_Forest"]
extra_trees_model = traditional_models["Extra_Trees"]

bagging_models = {
    "Random_Forest": random_forest_model,
    "Extra_Trees": extra_trees_model,
}

# Nhóm 3B: Boosting
adaboost_model = traditional_models["AdaBoost"]
gradient_boosting_model = traditional_models["Gradient_Boosting"]
xgboost_model = traditional_models["XGBoost"]
lightgbm_model = traditional_models["LightGBM"]
catboost_model = traditional_models["CatBoost"]

boosting_models = {
    "AdaBoost": adaboost_model,
    "Gradient_Boosting": gradient_boosting_model,
    "XGBoost": xgboost_model,
    "LightGBM": lightgbm_model,
    "CatBoost": catboost_model,
}

for subgroup_name, models in (("Bagging", bagging_models), ("Boosting", boosting_models)):
    print(f"\n{subgroup_name}")
    for model_name, model in models.items():
        model_path = f"{type(model).__module__}.{type(model).__name__}"
        print(f"[READY] {model_name}: {model_path}")
```

## Nhóm 4:Nhóm Học máy truyền thống khác dựa trên Xác suất & Khoảng cách
Đại điện các trường phái toán học phi tham số cơ bản khác

```python
# Nhóm 4: Mô hình dựa trên Xác suất và Khoảng cách
naive_bayes_model = traditional_models["Naive_Bayes"]
knn_model = traditional_models["KNN"]

group_4_models = {
    "Naive_Bayes": naive_bayes_model,
    "KNN": knn_model,
}

for model_name, model in group_4_models.items():
    model_path = f"{type(model).__module__}.{type(model).__name__}"
    print(f"[READY] {model_name}: {model_path}")
```
